### 🔁 Quick Reference: Choosing a file mode

Keep this in mind for every problem below.

```mermaid
flowchart TD
    A{"What do you need to do?"} -- "Read data that's already there" --> R["'r' mode<br/>FileNotFoundError if the file is missing"]
    A -- "Start a brand-new file<br/>(or replace an old one)" --> W["'w' mode<br/>⚠️ ERASES existing contents"]
    A -- "Add to the end of a file<br/>and keep what's there" --> AP["'a' mode<br/>creates the file if it's missing"]
```

**Reminders:** `write()` only takes strings (use `str()` or an f-string), and it doesn't add `'\n'` for you. Everything you read back is a string: strip it with `rstrip('\n')` or convert it with `int()` / `float()`.
