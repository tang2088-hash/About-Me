**Homework #1: Road Trip Playlist**

Scenario: You're making a playlist for a road trip and want to save it so it's still there tomorrow. Write **two void functions**:

* `write_playlist(filename, num_songs)` — opens the file in **write** mode and uses a `for` loop to ask the user for `num_songs` song titles, writing **each title on its own line**.
* `show_playlist(filename)` — opens the file in **read** mode and uses a `for` loop to print every song with a number in front of it. Use `rstrip('\n')` so there are **no blank lines** between songs.

In `main()`:
1. Ask the user how many songs they want, call `write_playlist`, then call `show_playlist`.
2. Ask the user for **one more** song and add it to the end of the file using **append** mode (don't erase the others!).
3. Call `show_playlist` again to show the updated playlist.

Example output (user input shown after the `?` and `:`):
```
How many songs? 3
Song 1: Life is a Highway
Song 2: Fast Car
Song 3: Take Me Home, Country Roads

Your playlist:
1. Life is a Highway
2. Fast Car
3. Take Me Home, Country Roads

Add one more song: Mr. Blue Sky

Your playlist:
1. Life is a Highway
2. Fast Car
3. Take Me Home, Country Roads
4. Mr. Blue Sky
```

````python
'''
Type Pseudocode here

FUNCTION write_playlist(filename, num_songs)
    OPEN filename in write mode
    FOR song_number FROM 1 THROUGH num_songs
        ASK user for song title
        WRITE song title then newline to the file
    END FOR
    CLOSE file
END FUNCTION

FUNCTION show_playlist(filename)
    SET song_number = 1
    DISPLAY "Your playlist:" 
    OPEN filename in read mode
    FOR each line in file
        REMOVE trailing newline
        DISPLAY song_number and song title
        ADD 1 to song_number
    END FOR
    CLOSE file
END FUNCTION

FUNCTION main()
    SET filename = "playlist.txt"
    ASK user how many songs they want
    CALL write_playlist(filename, num_songs)
    CALL show_playlist(filename)

    ASK user for one more song
    OPEN filename in append mode
    WRITE additional song followed by a newline
    CLOSE file

    CALL show_playlist(filename)
END FUNCTION

CALL main()

'''
# Type Code for Problem #1 Here!

def write_playlist(filename, num_songs):
    with open(filename, "w") as file:
        for song_number in range(1, num_songs + 1):
            title = input(f"Song {song_number}: ")
            file.write(f"{title}\n")

def show_playlist(filename):
    song_number = 1
    print("\nYour playlist:")
    
    with open(filename, "r") as file: 
        for line in file:
            title = line.rstrip("\n")
            print(f"{song_number}. {title}")
            song_number += 1

def main():
    filename = "playlist.txt"
    num_songs = int(input("How many songs? "))
    write_playlist(filename, num_songs)
    show_playlist(filename)

    extra_song = input("\nAdd one more song: ")
    with open(filename, "a") as file:
        file.write(f"{extra_song}\n")

    show_playlist(filename)

main()
````

````python
'''
Type Pseudocode here

FUNCTION analyze_temps(filename)
    SET days = 0
    SET total = 0

    OPEN filename in read mode
    FOR each line in file
        CONVERT line to a floating-point temperature

        IF days equals 0
            SET highest = temperature
            SET lowest = temperature
        ELSE
            IF temperature > highest
                SET highest = temperature
            END IF

            IF temperature < lowest
                SET lowest = temperature
            END IF
        END IF

        ADD temnperature to total
        ADD 1 to days
    END FOR
    CLOSE file

    SET average = total / days
    RETURN days, average, highest, lowest
END FUNCTION

FUNCTION write_report(report_filename, days, average, highest, lowest)
    OPEN report_filename in write mode
    WRITE days recorded on its own line
    WRITE average high with 1 decimal place(one per line)
    WRITE hottest day with 1 decimal place(one per line)
    WRITE coolest day with 1 decimal place(one per line)
    CLOSE file
END FUNCTION

FUNCTION main()
    CALL analyze_temp("temps.txt" and STORE four results
    CALL write_report("temp_report.txt", days, average, highest, lowest)

    OPEN "temp_report.txt" in read mode
    READ and DISPLAY its content
    CLOSE file
END FUNCTION

CALL main()

'''
# Type Code for Problem #2 Here!

def analyze_temps(filename):
    days = 0
    total = 0

    with open(filename, "r") as file:
        for line in file:
            temperature = float(line.strip())

            if days == 0:
                highest = temperature
                lowest = temperature
            else:
                if temperature > highest:
                    highest = temperature

                if temperature < lowest:
                    lowest = temperature

            total += temperature
            days += 1

    average = total / days
    return days, average, highest, lowest

def write_report(report_filename, days, average, highest, lowest):
    with open(report_filename, "w") as file:
        file.write(f"Days recorded: {days}\n")
        file.write(f"Average high: {average:.1f}\n")
        file.write(f"Hottest day: {highest:.1f}\n")
        file.write(f"Coolest day: {lowest:.1f}\n")

def main():
    days, average, highest, lowest = analyze_temps("temps.txt")
    write_report("temp_report.txt", days, average, highest, lowest)

    with open("temp_report.txt") as file: 
        print(file.read(), end="")
        
main()
````

Write two functions:

````python
'''
Type Pseudocode here

FUNCTION show_inventory(filename)
    SET total_value = 0
    DISPLAY column headings

    OPEN filename in read mode
    READ first item name

    WHILE item is not empty string
        REMOVE trailing newline from item
        READ quantity and CONVERT to integer
        READ price and CONVERT to float

        SET value = quantity * price
        ADD value to total_value
        DISPLAY item, quantity, price, and value in columns
        READ next item name
    END WHILE
    CLOSE file

    DISPLAY total inventory value
END FUNCTION

FUNCTION find_item(filename, search_item)
    OPEN filename in read mode
    READ first item name

    WHILE item is not an empty string
        REMOVE trailing newline from item
        READ quantity and CONVERT to integer
        READ price and CONVERT to float

        IF item = search_item
            RETURN quantity, price
        END IF
        READ next item name
    END WHILE
    CLOSE file

    RETURN None
END FUNCTION

FUNCTION main()
    CALL show_inventory("inventory.txt")
    ASK user for an item name
    SET result = find_item("inventory.txt", search_item)

    IF result is None
        DISPLAY item not found
    ELSE
        UNPACK result into quantity and price
        DISPLAY item name, quantity, and price
    END IF
END FUNCTION 

CALL main()

'''
# Type Code for Problem #3 Here!

def show_inventory(filename):
    total_value = 0
    print(f"{'item':<16}{'qty':>6}{'price':>10}{'value':>13}")

    with open(filename, "r") as file:
        item = file.readline()

        while item != '':
            item = item.rstrip('\n')
            qty = int(file.readline())
            price = float(file.readline())

            value = qty * price
            total_value += value
            print(f"{item:<16}{qty:>6} ${price:>7.2f} ${value:>10,.2f}")
            item = file.readline()

    print(f"Total inventory value: $ {total_value:,.2f}")

def find_item(filename, search_item):
    with open(filename, "r") as file:
        item = file.readline()

        while item != '':
            item = item.rstrip('\n')
            qty = int(file.readline())
            price = float(file.readline())

            if item == search_item:
                return qty, price
        
            item = file.readline()

    return None

def main():
    show_inventory("inventory.txt")
    search_item = input("\nType an item to look up: ")
    result = find_item("inventory.txt", search_item)

    if result is None:
        print(f"Sorry, {search_item} is not in the inventory.")
    else:
        qty, price = result
        print(f"{search_item}: {qty} in stock at ${price:.2f} each")

main()
````

**Homework #4: Crash-Proof Sales Totaler**

Scenario: A manager has a folder full of weekly sales files, but some of them contain typos (like `N/A` instead of a number) and sometimes the manager mistypes a filename. Your program must **never crash**.

Write a function `total_sales_file(filename)` that uses **all four** parts of a `try` statement:
* **`try`:** open the file and use a `for` loop to add up the amounts and count the lines.
* **`except FileNotFoundError`:** print a message that the file could not be found.
* **`except ValueError as err`:** print a message that the file has bad data, **including the default error message** stored in `err`.
* **`else`:** runs only if nothing went wrong: print the number of days, the total (with commas and 2 decimals), and the average per day.
  *(Notice the `\n` in the ValueError message below: the bad line still had its newline when `float()` tried to convert it.)*
* **`finally`:** always print `Finished processing <filename>.`

In `main()`, use a `while` loop so the user can check **as many files as they want** (keep asking until they type `done`). Test it with `sales_week1.txt`, `sales_week2.txt`, and a filename that doesn't exist.

Example output:
```
Enter a sales file name (or 'done' to quit): sales_week1.txt
Days: 5
Total sales: $6,386.80
Average per day: $1,277.36
Finished processing sales_week1.txt.

Enter a sales file name (or 'done' to quit): sales_week2.txt
sales_week2.txt contains bad data: could not convert string to float: 'N/A\n'
Finished processing sales_week2.txt.

Enter a sales file name (or 'done' to quit): sales_week3.txt
Error: sales_week3.txt could not be found.
Finished processing sales_week3.txt.

Enter a sales file name (or 'done' to quit): done
Goodbye!
```

````python
'''
Type Pseudocode here

FUNCTION total_sales_file(filename)
    SET total = 0.0
    SET days = 0

    TRY
        OPEN filename in read mode
        FOR each line in file
            CONVERT line to float
            ADD amount to total
            ADD 1 to days
        END FOR
        CLOSE file

    EXCEPT FileNotFoundError
        DISPLAY file could not be found message

    EXCEPT ValueError as err
        DISPLAY bad data message including err

    ELSE
        IF days > 0 
            SET average = total / days
        ELSE
            SET average = 0.0
        END IF

        DISPLAY days, total, and average

    FINALLY
        DISPLAY "Finished processing" and filename
    END TRY
END FUNCTION

FUNCTION main()
    ASK user for filename or "done"

    WHILE filename is not "done"
        CALL total_sales_file(filename)
        ASK user for another filename or "done" 
    END WHILE

    DISPLAY "Goodbye!"
END FUNCTION

CALL main()

'''
# Type Code for Problem #4 Here!

def total_sales_file(filename):
    total = 0.0
    days = 0

    try:
        with open(filename, "r") as file:
            for line in file:
                amount = float(line)
                total += amount          
                days += 1

    except FileNotFoundError:
        print(f"Error: {filename} could not be found.")

    except ValueError as err:
        print(f"{filename} contains bad data: {err}")

    else:
        if days > 0:
            average = total / days
        else:
            average = 0.0

        print(f"Days: {days}")
        print(f"Total Sales: ${total:,.2f}")
        print(f"Average per day: ${average:,.2f}")

    finally:
        print(f"Finished processing {filename}.")

def main():
    filename = input("Enter a sales file name (or 'done' to quit): ")

    while filename != "done":
        total_sales_file(filename)
        filename = input(
            "\nEnter a sales file name (or 'done' to quit): "
        )
        
    print("Goodbye!")
    
main()
````

**Homework #5 (Challenge): Personal Expense Tracker**

Scenario: You want to track your spending across the semester. Every expense is saved as a **record** with **two fields** in `expenses.txt`: a **description** and an **amount**. Because the data is in a file, your expenses are still there the next time you run the program!

Build the program from these functions:

* A global constant `FILENAME = 'expenses.txt'` at the top of the program.
* `show_menu()` — a **void** function that prints the menu:
  ```
  1. Add an expense
  2. View all expenses
  3. Show summary
  4. Quit
  ```
* `get_amount(prompt)` — asks the user the question in `prompt` and uses a `while` loop with **`try`/`except ValueError`** to **keep asking** until the user enters a valid number that is **greater than 0**. **Returns** the amount as a `float`.
* `add_expense(filename)` — asks for a description, calls `get_amount()`, and **appends** the record (two lines) to the file.
* `view_expenses(filename)` — reads and prints every record in lined-up columns. If the file doesn't exist yet, catch the **`FileNotFoundError`** and print `No expenses recorded yet.` instead of crashing.
* `summarize(filename)` — **returns three values**: the **number of expenses**, the **total spent**, and the **largest expense** amount. If the file doesn't exist, return `0, 0.0, 0.0`.
* `main()` — shows the menu in a `while` loop until the user chooses `4`. Print `Invalid choice.` for anything other than 1–4. For choice 3, print the summary (money with commas and 2 decimals, plus the **average** expense — watch out for dividing by zero when there are no expenses!).

Example run (starting with **no** `expenses.txt` file):

💡 **Testing tip:** To start over with an empty tracker, delete `expenses.txt` from the Files panel, or run `open('expenses.txt', 'w').close()` in a new cell.

````python
'''
Type Pseudocode here

SET global constant FILENAME = "expenses.txt"

FUNCTION show_menu()
    DISPLAY "1. Add an expense"
    DISPLAY "2. View all expenses"
    DISPLAY "3. Show summary"
    DISPLAY "4. Quit"
END FUNCTION

FUNCTION get_amount(prompt)
    WHILE True
        TRY
            ASK user using prompt and CONVERT input to float

            IF amount > 0
                RETURN amount
            ELSE
                DISPLAY "The amount must be greater than 0."
            END IF

        EXCEPT ValueError
            DISPLAY "Please enter a valid number."
        END TRY
    END WHILE
END FUNCTION

FUNCTION add_expense(filename)
    ASK user for description
    SET amount = get_amount("Amount: $")

    OPEN filename in append mode
    WRITE description followed by newline
    WRITE amount followed by newline
    CLOSE file

    DISPLAY saved expense
END FUNCTION

FUNCTION view_expenses(filename)
    TRY
        OPEN filename in read mode
        READ first description

        WHILE description is not empty string
            REMOVE trailing newline from description
            READ amount and CONVERT to float
            DISPLAY description and amount in aligned columns
            READ next description
        END WHILE
        CLOSE file

    EXCEPT FileNotFoundError
        DISPLAY "No expenses recorded yet."
    END TRY
END FUNCTION

FUNCTION summarize(filename)
    SET count = 0
    SET total = 0.0
    SET largest = 0.0

    TRY
        OPEN filename in read mode
        READ first description

        WHILE description is not empty string
            READ amount and CONVERT to float
            ADD 1 to count
            ADD amount to total

            IF amount > largest
                SET largest = amount
            END IF

            READ next description
        END WHILE
        CLOSE file

    EXCEPT FileNotFoundError
        RETURN 0, 0.0, 0.0
    END TRY

    RETURN count, total, largest
END FUNCTION

FUNCTION main()
    SET choice = ""

    WHILE choice is not "4"
        CALL show_menu()
        ASK user for choice

        IF choice equals "1"
            CALL add_expense(FILENAME)
        ELSE IF choice equals "2"
            CALL view_expenses(FILENAME)
        ELSE IF choice equals "3"
            SET count, total, largest = summarize(FILENAME)

            IF count > 0
                SET average = total / count
            ELSE
                SET average = 0.0
            END IF

            DISPLAY count, total, average, and largest
        ELSE IF choice equals "4"
            DISPLAY "Goodbye!"
        ELSE
            DISPLAY "Invalid choice."
        END IF
    END WHILE
END FUNCTION

CALL main()
'''
# Type Code for Problem #5 Here!

FILENAME = "expenses.txt"

def show_menu():
    print("1. Add an expense")
    print("2. View all expenses")
    print("3. Show summary")
    print("4. Quit")

def get_amount(prompt):
    while True:
        try:
            amount = float(input(prompt))

            if amount > 0:
                return amount
            else:
                print("The amount must be greater than 0.")

        except ValueError:
            print("Please enter a valid number.")

def add_expense(filename):
    description = input("Description: ")
    amount = get_amount("Amount: $")

    with open(filename, "a") as file:
        file.write(f"{description}\n")
        file.write(f"{amount}\n")
        
    print(f"Saved: {description} ${amount:,.2f}")

def view_expenses(filename):
    try:
        with open(filename, "r") as file:
            description = file.readline()

            while description != '':
                description = description.rstrip('\n')
                amount = float(file.readline())
                print(f"{description:<24} ${amount:>8,.2f}")
                description = file.readline()

    except FileNotFoundError:
        print("No expenses recorded yet.")

def summarize(filename):
    count = 0
    total = 0.0
    largest = 0.0

    try:
        with open(filename, "r") as file:
            description = file.readline()

            while description != '':
                amount = float(file.readline())
                count += 1
                total += amount

                if amount > largest:
                    largest = amount

                description = file.readline()

    except FileNotFoundError:
        return 0, 0.0, 0.0

    return count, total, largest

def main():
    choice = ""

    while choice != "4":
        show_menu()

        choice = input("Enter your choice: ")

        if choice == "1":
            add_expense(FILENAME)
        elif choice == "2":
            view_expenses(FILENAME)
        elif choice == "3":
            count, total, largest = summarize(FILENAME)

            if count > 0:
                average = total / count
            else:
                average = 0.0

            print(f"Expenses: {count}")
            print(f"Total Spent: ${total:,.2f}")
            print(f"Average expense: ${average:,.2f}")
            print(f"Largest expense: ${largest:,.2f}")
        elif choice == "4":
            print("Goodbye!")
        else:
            print("Invalid choice.")
        print()

main()
````
